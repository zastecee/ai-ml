# Combine the folder path and the specific file name
file_path = os.path.join(path, 'melb_data.csv')

# Load the CSV file into pandas
melbourne_data = pd.read_csv(file_path)

```python
# print a summary of the data in Melbourne data
melbourne_data.describe()
```

```python
# Your First Machine Learning Model

melbourne_data.columns
```

```python
melbourne_data = melbourne_data.dropna(axis=0)
```

```python
# Selecting The Prediction Target
y = melbourne_data.Price
```

```python
# Choosing "Features"
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']

X = melbourne_data[melbourne_features]

X.describe()
```

```python
X.head()
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
# pip install scikit-learn
```

```python
# Define model. Specify a number for random_state to ensure same results each run
melbourne_model = DecisionTreeRegressor(random_state=0)
```

```python
# Fit model
melbourne_model.fit(X, y)
```

```python
print("Making predictions for the following 5 houses:")
print(X.head())
print("The predictions are")
print(melbourne_model.predict(X.head()))
```

```python
predictions = melbourne_model.predict(X)
```

```python
comparison_df = pd.DataFrame({
    "A": predictions,
    "B": y
})

comparison_df
```

```python
from sklearn.metrics import mean_absolute_error

predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y, predicted_home_prices)
```

```python
# The Proble with "In-Sample" Scores
from sklearn.model_selection import train_test_split

# split data into training and validation data, for both features and target
# The split is based on a random number generator. Supplying a numeric value to
# the random_state argument guarantees we get the same split every time we
# run this script.
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state=0)

# Define model
melbourne_model_2 = DecisionTreeRegressor(random_state=0)

# Fit model
melbourne_model_2.fit(train_X, train_y)

# get predicted prices on validation data
val_predictions = melbourne_model_2.predict(val_X)
mean_absolute_error(val_y, val_predictions)

# There are many ways to improve this model, such as experimenting to find better features or different model types.
```
